```python
from sklearn.metrics import confusion_matrix

y_true = [2, 0, 2, 2, 0, 1]
y_pred = [0, 0, 2, 2, 0, 2]
sklearn.metrics.confusion_matrix(y_true, y_pred)
    
y_true = ["cat", "ant", "cat", "cat", "ant", "bird"]
y_pred = ["ant", "ant", "cat", "cat", "ant", "cat"]
sklearn.metrics.confusion_matrix(y_true, y_pred, labels=["ant", "bird", "cat"])

tn, fp, fn, tp = sklearn.metrics.confusion_matrix([0, 1, 0, 1], [1, 1, 1, 0]).ravel().tolist()
(tn, fp, fn, tp)
```
