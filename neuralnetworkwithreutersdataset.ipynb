```python
# ============================================
# IMPORTS
# ============================================

import numpy as np
import matplotlib.pyplot as plt
import keras

from keras import layers
from keras.optimizers import SGD
from keras.datasets import reuters
from keras.utils import to_categorical


# ============================================
# LOAD REUTERS DATASET
# ============================================

(x_train, y_train), (x_test, y_test) = reuters.load_data(
    num_words=10000,
    test_split=0.3,
    seed=42
)

print(f"X Train length: {len(x_train)}")
print(f"y Train length: {len(y_train)}")
print(f"X Test length: {len(x_test)}")
print(f"y Test length: {len(y_test)}")

print(f"Total data examples: {len(x_train) + len(x_test)}")

print(x_train[0])
print(y_train[0])


# ============================================
# VIEW VOCABULARY
# ============================================

vocabulary = reuters.get_word_index()

# Invert the mapping.
# Numeric indices become keys and words become values.
vocabulary = {v: k for k, v in vocabulary.items()}

for k, v in vocabulary.items():
    print(k, v)

for word_idx in x_train[0]:
    print(vocabulary.get(word_idx - 3, '?'))


# ============================================
# PRE-PROCESS TRAINING DATA
# ============================================

x_train_enc = np.zeros(
    shape=(len(x_train), 10000)
)

print("Encoded training data shape:", x_train_enc.shape)

for row_number, word_idx_seq in enumerate(x_train):
    # Set the relevant word indices to 1.
    x_train_enc[row_number, word_idx_seq] = 1

print(x_train_enc[0])


# ============================================
# PRE-PROCESS TEST DATA
# ============================================

x_test_enc = np.zeros(
    shape=(len(x_test), 10000)
)

print("Encoded test data shape:", x_test_enc.shape)

for row_number, word_idx_seq in enumerate(x_test):
    # Set the relevant word indices to 1.
    x_test_enc[row_number, word_idx_seq] = 1

print(x_test_enc[0])


# ============================================
# CONVERT LABELS TO CATEGORICAL FORMAT
# ============================================

y_train_enc = to_categorical(y_train)
y_test_enc = to_categorical(y_test)

print("Encoded training labels shape:", y_train_enc.shape)
print("Encoded test labels shape:", y_test_enc.shape)


# ============================================
# CREATE VALIDATION SET
# ============================================

x_val = x_train_enc[:1000]
y_val = y_train_enc[:1000]

x_train_enc_rest = x_train_enc[1000:]
y_train_enc_rest = y_train_enc[1000:]

print("Training data:", x_train_enc_rest.shape)
print("Validation data:", x_val.shape)
print("Test data:", x_test_enc.shape)


# ============================================
# 1. BASELINE MODEL
# ============================================

baseline_model = keras.Sequential()

baseline_model.add(
    layers.Input(shape=(10000,))
)

baseline_model.add(
    layers.Dense(64, activation="relu")
)

baseline_model.add(
    layers.Dense(64, activation="relu")
)

baseline_model.add(
    layers.Dense(46, activation="softmax")
)

baseline_model.summary()


baseline_model.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


# ============================================
# TRAIN BASELINE MODEL
# ============================================

baseline_history = baseline_model.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=120,
    validation_data=(x_val, y_val)
)


# ============================================
# 2. BASELINE TRAINING / VALIDATION LOSS
# ============================================

plt.figure(figsize=(10, 6))

plt.plot(
    baseline_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    baseline_history.history["val_loss"],
    label="Validation Loss"
)

plt.axvline(
    x=59,
    linestyle="--",
    label="Possible stopping point"
)

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Baseline Training and Validation Loss")

plt.legend()
plt.show()


# ============================================
# 3/4. FRESH BASELINE MODEL
# STOPPING AT 60 EPOCHS
# ============================================

baseline_final = keras.Sequential()

baseline_final.add(
    layers.Input(shape=(10000,))
)

baseline_final.add(
    layers.Dense(64, activation="relu")
)

baseline_final.add(
    layers.Dense(64, activation="relu")
)

baseline_final.add(
    layers.Dense(46, activation="softmax")
)


baseline_final.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


baseline_final_history = baseline_final.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=60,
    validation_data=(x_val, y_val)
)


# ============================================
# 5. BASELINE TEST EVALUATION
# ============================================

baseline_test_loss, baseline_test_accuracy = baseline_final.evaluate(
    x_test_enc,
    y_test_enc,
    verbose=0
)

print("Baseline Test Loss:", baseline_test_loss)
print("Baseline Test Accuracy:", baseline_test_accuracy)


baseline_predictions = baseline_final.predict(
    x_test_enc
)

baseline_predicted_classes = np.argmax(
    baseline_predictions,
    axis=1
)

print(
    "First predicted class:",
    baseline_predicted_classes[0]
)

print(
    "Actual class:",
    y_test[0]
)


# ============================================
# 6. MODEL 2 - 4 NEURONS IN HIDDEN LAYERS
# ============================================

model2 = keras.Sequential()

model2.add(
    layers.Input(shape=(10000,))
)

model2.add(
    layers.Dense(4, activation="relu")
)

model2.add(
    layers.Dense(4, activation="relu")
)

model2.add(
    layers.Dense(46, activation="softmax")
)

model2.summary()


model2.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


model2_history = model2.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=120,
    validation_data=(x_val, y_val)
)


# ============================================
# MODEL 2 LOSS
# ============================================

plt.figure(figsize=(10, 6))

plt.plot(
    model2_history.history["loss"],
    label="Model 2 Training Loss"
)

plt.plot(
    model2_history.history["val_loss"],
    label="Model 2 Validation Loss"
)

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Model 2 Training and Validation Loss")

plt.legend()
plt.show()


# ============================================
# MODEL 2 TEST EVALUATION
# ============================================

model2_test_loss, model2_test_accuracy = model2.evaluate(
    x_test_enc,
    y_test_enc,
    verbose=0
)

print("Model 2 Test Loss:", model2_test_loss)
print("Model 2 Test Accuracy:", model2_test_accuracy)


# ============================================
# 7/8. MODEL 3 - DROPOUT 0.2
# ============================================

model3 = keras.Sequential()

model3.add(
    layers.Input(shape=(10000,))
)

model3.add(
    layers.Dense(64, activation="relu")
)

model3.add(
    layers.Dropout(0.2)
)

model3.add(
    layers.Dense(64, activation="relu")
)

model3.add(
    layers.Dropout(0.2)
)

model3.add(
    layers.Dense(46, activation="softmax")
)

model3.summary()


model3.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


model3_history = model3.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=240,
    validation_data=(x_val, y_val)
)


# ============================================
# MODEL 4 - DROPOUT 0.5
# ============================================

model4 = keras.Sequential()

model4.add(
    layers.Input(shape=(10000,))
)

model4.add(
    layers.Dense(64, activation="relu")
)

model4.add(
    layers.Dropout(0.5)
)

model4.add(
    layers.Dense(64, activation="relu")
)

model4.add(
    layers.Dropout(0.5)
)

model4.add(
    layers.Dense(46, activation="softmax")
)

model4.summary()


model4.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


model4_history = model4.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=240,
    validation_data=(x_val, y_val)
)


# ============================================
# DROPOUT LOSS COMPARISON
# ============================================

plt.figure(figsize=(12, 7))

plt.plot(
    model3_history.history["loss"],
    "--",
    label="Dropout 0.2 - Training"
)

plt.plot(
    model3_history.history["val_loss"],
    label="Dropout 0.2 - Validation"
)

plt.plot(
    model4_history.history["loss"],
    "--",
    label="Dropout 0.5 - Training"
)

plt.plot(
    model4_history.history["val_loss"],
    label="Dropout 0.5 - Validation"
)

plt.plot(
    baseline_history.history["loss"],
    "--",
    label="Baseline - Training"
)

plt.plot(
    baseline_history.history["val_loss"],
    label="Baseline - Validation"
)

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Baseline vs Dropout Training and Validation Loss")

plt.legend()
plt.show()


# ============================================
# 9. TEST ACCURACY COMPARISON
# ============================================

baseline_loss, baseline_accuracy = baseline_final.evaluate(
    x_test_enc,
    y_test_enc,
    verbose=0
)

model3_loss, model3_accuracy = model3.evaluate(
    x_test_enc,
    y_test_enc,
    verbose=0
)

model4_loss, model4_accuracy = model4.evaluate(
    x_test_enc,
    y_test_enc,
    verbose=0
)


print("Baseline accuracy:", baseline_accuracy)
print("Model 3 accuracy:", model3_accuracy)
print("Model 4 accuracy:", model4_accuracy)


# ============================================
# ACCURACY COMPARISON
# ============================================

models = [
    "Baseline",
    "Dropout 0.2",
    "Dropout 0.5"
]

accuracies = [
    baseline_accuracy,
    model3_accuracy,
    model4_accuracy
]


plt.figure(figsize=(8, 6))

plt.bar(
    models,
    accuracies
)

plt.ylabel("Test Accuracy")
plt.xlabel("Model")
plt.title("Test Accuracy Comparison")

plt.ylim(0, 1)

plt.show()


# ============================================
# TRAINING ACCURACY COMPARISON
# ============================================

plt.figure(figsize=(10, 6))

plt.plot(
    baseline_history.history["accuracy"],
    label="Baseline"
)

plt.plot(
    model3_history.history["accuracy"],
    label="Dropout 0.2"
)

plt.plot(
    model4_history.history["accuracy"],
    label="Dropout 0.5"
)

plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("Training Accuracy Comparison")

plt.legend()
plt.show()


# ============================================
# 10. BATCH NORMALISATION MODEL
# ============================================

bn_model = keras.Sequential()

bn_model.add(
    layers.Input(shape=(10000,))
)

bn_model.add(
    layers.Dense(64)
)

bn_model.add(
    layers.BatchNormalization()
)

bn_model.add(
    layers.ReLU()
)

bn_model.add(
    layers.Dense(64)
)

bn_model.add(
    layers.BatchNormalization()
)

bn_model.add(
    layers.ReLU()
)

bn_model.add(
    layers.Dense(46, activation="softmax")
)

bn_model.summary()


bn_model.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


bn_history = bn_model.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=150,
    validation_data=(x_val, y_val)
)


# ============================================
# BATCH NORMALISATION LOSS
# ============================================

plt.figure(figsize=(10, 6))

plt.plot(
    bn_history.history["loss"],
    "--",
    label="BN Training Loss"
)

plt.plot(
    bn_history.history["val_loss"],
    label="BN Validation Loss"
)

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Batch Normalisation Training and Validation Loss")

plt.legend()
plt.show()


# ============================================
# 11. BATCH NORMALISATION + L2
# ============================================

bn_l2_model = keras.Sequential()

bn_l2_model.add(
    layers.Input(shape=(10000,))
)

bn_l2_model.add(
    layers.Dense(
        64,
        kernel_regularizer=keras.regularizers.l2()
    )
)

bn_l2_model.add(
    layers.BatchNormalization()
)

bn_l2_model.add(
    layers.ReLU()
)

bn_l2_model.add(
    layers.Dense(
        64,
        kernel_regularizer=keras.regularizers.l2()
    )
)

bn_l2_model.add(
    layers.BatchNormalization()
)

bn_l2_model.add(
    layers.ReLU()
)

bn_l2_model.add(
    layers.Dense(
        46,
        activation="softmax"
    )
)

bn_l2_model.summary()


bn_l2_model.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


bn_l2_history = bn_l2_model.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=150,
    validation_data=(x_val, y_val)
)


# ============================================
# BN VS BN + L2 LOSS
# ============================================

plt.figure(figsize=(12, 7))

plt.plot(
    bn_history.history["loss"],
    "--",
    label="BN Training"
)

plt.plot(
    bn_history.history["val_loss"],
    label="BN Validation"
)

plt.plot(
    bn_l2_history.history["loss"],
    "--",
    label="BN + L2 Training"
)

plt.plot(
    bn_l2_history.history["val_loss"],
    label="BN + L2 Validation"
)

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Batch Normalisation vs Batch Normalisation + L2")

plt.legend()
plt.show()


# ============================================
# 12. RELU VARIANTS
# ============================================

print("Available ReLU variants used in this experiment:")

print("1. LeakyReLU")
print("2. ELU - Exponential Linear Unit")
print("3. PReLU - Parametric ReLU")
print("4. GELU - Gaussian Error Linear Unit")


# ============================================
# 13. MODEL 5 - LEAKY RELU
# ============================================

model5 = keras.Sequential()

model5.add(
    layers.Input(shape=(10000,))
)

model5.add(
    layers.Dense(64)
)

model5.add(
    layers.LeakyReLU()
)

model5.add(
    layers.Dense(64)
)

model5.add(
    layers.LeakyReLU()
)

model5.add(
    layers.Dense(46, activation="softmax")
)


model5.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


model5_history = model5.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=120,
    validation_data=(x_val, y_val)
)


# ============================================
# MODEL 6 - ELU
# ============================================

model6 = keras.Sequential()

model6.add(
    layers.Input(shape=(10000,))
)

model6.add(
    layers.Dense(64)
)

model6.add(
    layers.ELU()
)

model6.add(
    layers.Dense(64)
)

model6.add(
    layers.ELU()
)

model6.add(
    layers.Dense(46, activation="softmax")
)


model6.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


model6_history = model6.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=120,
    validation_data=(x_val, y_val)
)


# ============================================
# MODEL 7 - PRELU
# ============================================

model7 = keras.Sequential()

model7.add(
    layers.Input(shape=(10000,))
)

model7.add(
    layers.Dense(64)
)

model7.add(
    layers.PReLU()
)

model7.add(
    layers.Dense(64)
)

model7.add(
    layers.PReLU()
)

model7.add(
    layers.Dense(46, activation="softmax")
)


model7.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


model7_history = model7.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=120,
    validation_data=(x_val, y_val)
)


# ============================================
# MODEL 8 - GELU
# ============================================

model8 = keras.Sequential()

model8.add(
    layers.Input(shape=(10000,))
)

model8.add(
    layers.Dense(64)
)

model8.add(
    layers.Activation("gelu")
)

model8.add(
    layers.Dense(64)
)

model8.add(
    layers.Activation("gelu")
)

model8.add(
    layers.Dense(46, activation="softmax")
)


model8.compile(
    loss="categorical_crossentropy",
    optimizer=SGD(learning_rate=0.05),
    metrics=["accuracy"]
)


model8_history = model8.fit(
    x_train_enc_rest,
    y_train_enc_rest,
    batch_size=512,
    epochs=120,
    validation_data=(x_val, y_val)
)


# ============================================
# MODELS 5-8 VALIDATION LOSS COMPARISON
# ============================================

plt.figure(figsize=(12, 7))

plt.plot(
    model5_history.history["val_loss"],
    label="Model 5 - LeakyReLU"
)

plt.plot(
    model6_history.history["val_loss"],
    label="Model 6 - ELU"
)

plt.plot(
    model7_history.history["val_loss"],
    label="Model 7 - PReLU"
)

plt.plot(
    model8_history.history["val_loss"],
    label="Model 8 - GELU"
)

plt.xlabel("Epochs")
plt.ylabel("Validation Loss")
plt.title("Validation Loss Comparison of Activation Functions")

plt.legend()
plt.show()


# ============================================
# MODELS 5-8 TRAINING LOSS COMPARISON
# ============================================

plt.figure(figsize=(12, 7))

plt.plot(
    model5_history.history["loss"],
    "--",
    label="Model 5 - LeakyReLU"
)

plt.plot(
    model6_history.history["loss"],
    "--",
    label="Model 6 - ELU"
)

plt.plot(
    model7_history.history["loss"],
    "--",
    label="Model 7 - PReLU"
)

plt.plot(
    model8_history.history["loss"],
    "--",
    label="Model 8 - GELU"
)

plt.xlabel("Epochs")
plt.ylabel("Training Loss")
plt.title("Training Loss Comparison of Activation Functions")

plt.legend()
plt.show()


# ============================================
# MODELS 5-8 TEST ACCURACY
# ============================================

model5_loss, model5_accuracy = model5.evaluate(
    x_test_enc,
    y_test_enc,
    verbose=0
)

model6_loss, model6_accuracy = model6.evaluate(
    x_test_enc,
    y_test_enc,
    verbose=0
)

model7_loss, model7_accuracy = model7.evaluate(
    x_test_enc,
    y_test_enc,
    verbose=0
)

model8_loss, model8_accuracy = model8.evaluate(
    x_test_enc,
    y_test_enc,
    verbose=0
)


print("Model 5 - LeakyReLU:", model5_accuracy)
print("Model 6 - ELU:", model6_accuracy)
print("Model 7 - PReLU:", model7_accuracy)
print("Model 8 - GELU:", model8_accuracy)


# ============================================
# ACTIVATION FUNCTION ACCURACY COMPARISON
# ============================================

activation_models = [
    "LeakyReLU",
    "ELU",
    "PReLU",
    "GELU"
]

activation_accuracy = [
    model5_accuracy,
    model6_accuracy,
    model7_accuracy,
    model8_accuracy
]


plt.figure(figsize=(9, 6))

plt.bar(
    activation_models,
    activation_accuracy
)

plt.xlabel("Activation Function")
plt.ylabel("Test Accuracy")
plt.title("Test Accuracy by Activation Function")

plt.ylim(0, 1)

plt.show()


# ============================================
# FINAL MODEL COMPARISON
# ============================================

results = {
    "Baseline": baseline_accuracy,
    "Model 2 (4 neurons)": model2_test_accuracy,
    "Model 3 (Dropout 0.2)": model3_accuracy,
    "Model 4 (Dropout 0.5)": model4_accuracy,
    "Model 5 (LeakyReLU)": model5_accuracy,
    "Model 6 (ELU)": model6_accuracy,
    "Model 7 (PReLU)": model7_accuracy,
    "Model 8 (GELU)": model8_accuracy
}


for model_name, accuracy in results.items():
    print(f"{model_name}: {accuracy:.4f}")

```
